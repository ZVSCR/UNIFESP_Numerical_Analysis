```python
import math
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
def truncar(numero, casas):
  # Calcula o fator que desloca a vírgula decimal.
  fator = 10 ** casas

  # Trunca o número deslocado e devolve a vírgula à posição original.
  return math.trunc(numero * fator) / fator
```

```python
valores = [3.14159265, 98.76543, 0.00498765]

casas = [0, 1, 2, 3, 4]

Erro_absoluto_trunc = []
Erro_absoluto_round = []

dados = []

for valor in valores:
  print("-"*20)
  for n in casas:
    absoluto_truncamento = abs(valor - truncar(valor, n))
    relativo_truncamento = absoluto_truncamento/abs(valor)
    percentual_truncamento = 100 * relativo_truncamento

    absoluto_round = abs(valor - round(valor, n))
    relativo_round = absoluto_round/abs(valor)
    percentual_round = 100 * relativo_round

    print(f'truncamento absoluto: {absoluto_truncamento}')
    print(f'truncamento relativo: {relativo_truncamento}')
    print(f'truncamento percentual: {percentual_truncamento}')

    print(f'round absoluto: {absoluto_round}')
    print(f'round relativo: {relativo_round}')
    print(f'round percentual: {percentual_round}')
    print("\n")

    dados.append({
      "Valor": valor,
      "Casas": n,

      "Erro Abs. Truncamento": absoluto_truncamento,
      "Erro Rel. Truncamento": relativo_truncamento,
      "Erro % Truncamento": percentual_truncamento,

      "Erro Abs. Round": absoluto_round,
      "Erro Rel. Round": relativo_round,
      "Erro % Round": percentual_round
    })

    Erro_absoluto_trunc.append(absoluto_truncamento)
    Erro_absoluto_round.append(absoluto_round)
```

```python
# Tabela

df = pd.DataFrame(dados)

df
```

```python
plt.figure(figsize=(10, 6))

# Uma curva para cada valor
for valor in df["Valor"].unique():
    dados = df[df["Valor"] == valor]

    plt.plot(
        dados["Casas"],
        dados["Erro Abs. Truncamento"],
        marker="o",
        linewidth=2,
        label=f"Truncamento ({valor})"
    )

    plt.plot(
        dados["Casas"],
        dados["Erro Abs. Round"],
        marker="s",
        linestyle="--",
        linewidth=2,
        label=f"Round ({valor})"
    )

plt.yscale("log")  # escala logarítmica no eixo Y

plt.xlabel("Número de casas decimais (n)")
plt.ylabel("Erro absoluto")
plt.title("Erro absoluto de truncamento e arredondamento para os 3 valores")
plt.grid(True, which="both", linestyle=":")
plt.legend(fontsize=8)
plt.tight_layout()

plt.show()
```

# Parte B - Medidasa de pressão Arterial

```python
pressao = [121.7, 120.4, 122.1, 119.8, 121.2, 120.9, 122.4, 121.0, 120.2, 121.5]
pressao_round = []
pressao_truncamento = []
for p in pressao:
  pressao_round.append(round(p, 0))
  pressao_truncamento.append(truncar(p, 0))
```

```python
#Media

ave_round = sum(pressao_round)/len(pressao_round)
ave_truncamento = sum(pressao_truncamento)/len(pressao_truncamento)
ave_pressao = sum(pressao)/len(pressao)

print(f'Media round: {ave_round}')
print(f'Media truncamento: {ave_truncamento}')
print(f'Media pressao: {ave_pressao}')

# Min

min_pressao = min(pressao)
min_round = min(pressao_round)
min_truncamento = min(pressao_truncamento)

print("\n=== MÍNIMO ===")
print(f"Pressão:       {min_pressao}")
print(f"Round:         {min_round}")
print(f"Truncamento:   {min_truncamento}")

# Max
max_pressao = max(pressao)
max_round = max(pressao_round)
max_truncamento = max(pressao_truncamento)

print("\n=== MÁXIMO ===")
print(f"Pressão:       {max_pressao}")
print(f"Round:         {max_round}")
print(f"Truncamento:   {max_truncamento}")


# AMPLITUDE

amplitude_pressao = max_pressao - min_pressao
amplitude_round = max_round - min_round
amplitude_truncamento = max_truncamento - min_truncamento

print("\n=== AMPLITUDE ===")
print(f"Pressão:       {amplitude_pressao}")
print(f"Round:         {amplitude_round}")
print(f"Truncamento:   {amplitude_truncamento}")



# DESVIO PADRÃO

def desvio_padrao(dados):
    media = sum(dados) / len(dados)
    variancia = sum((x - media) ** 2 for x in dados) / len(dados)
    return math.sqrt(variancia)

dp_pressao = desvio_padrao(pressao)
dp_round = desvio_padrao(pressao_round)
dp_truncamento = desvio_padrao(pressao_truncamento)

print("\n=== DESVIO PADRÃO ===")
print(f"Pressão:       {dp_pressao}")
print(f"Round:         {dp_round}")
print(f"Truncamento:   {dp_truncamento}")

```
